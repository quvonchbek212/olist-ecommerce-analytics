# Which products are most beneficial?


```python
# the 10 most profitable products
query = """
SELECT product_id,
SUM(price) AS total_price,
SUM(freight_value) AS total_freight
FROM order_items
GROUP BY product_id
ORDER BY total_price DESC
LIMIT 10;
"""
pd.read_sql(query, conn)
```

# Which customers are most profitable?


```python
# 20 most profitable customers

query = """
SELECT o.customer_id,
SUM(oi.price) AS total_spent
FROM orders AS o
JOIN order_items AS oi USING(order_id)
GROUP BY o.customer_id
ORDER BY total_spent DESC
LIMIT 20;
"""
pd.read_sql(query, conn)
```

# Which regions generate the most revenue?


```python
# 20 most profitable States

query = """
SELECT c.customer_state,
SUM(oi.price)/1000000 AS revenue
FROM orders o
JOIN customers c USING(customer_id)
JOIN order_items oi USING(order_id)
GROUP BY c.customer_state
ORDER BY revenue DESC;
"""
pd.read_sql(query, conn)
```

# Which category is the most useful?

```python
# Category-level profit

query = """
SELECT p.product_category_name,
SUM(oi.price) as revenue,
SUM(oi.freight_value) AS cost,
SUM(oi.price - oi.freight_value) AS profit
FROM order_items oi
JOIN products p USING(product_id)
GROUP BY p.product_category_name
ORDER BY profit DESC;
"""
pd.read_sql(query, conn)
```

# Where is slow delivery causing the most damage?

```python
query = """
SELECT p.product_category_name,
AVG(julianday(order_delivered_customer_date) - julianday(order_purchase_timestamp)) AS delivery_days
FROM orders o
JOIN order_items oi USING(order_id)
JOIN products p USING(product_id)
JOIN order_reviews r USING(order_id)
WHERE  o.order_delivered_customer_date IS NOT NULL
GROUP BY p.product_category_name
ORDER BY delivery_days DESC;
"""
pd.read_sql(query, conn)
```
